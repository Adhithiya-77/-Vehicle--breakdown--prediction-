```python
import pandas as pd
import numpy as np

df = pd.read_csv("synthetic_telemetry_data.csv")

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nUnique Values:")
print(df.nunique())

print("\nBasic Statistics:")
display(df.describe().T)

print("\nObject/Categorical Columns:")
print(df.select_dtypes(include=["object"]).columns.tolist())

print("\nNumerical Columns:")
print(df.select_dtypes(include=[np.number]).columns.tolist())
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["failure_date"] = pd.to_datetime(df["failure_date"])

df["breakdown_within_30days"] = (
    (df["failure_date"] >= df["timestamp"]) &
    (df["failure_date"] <= df["timestamp"] + pd.Timedelta(days=30))
).astype(int)

print("Target Distribution:")
print(df["breakdown_within_30days"].value_counts())

print("\nTarget Percentage:")
print(
    df["breakdown_within_30days"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
drop_cols = [
    "failure_date",
    "failure_type",
    "engine_failure_imminent",
    "brake_issue_imminent",
    "battery_issue_imminent",
    "timestamp",
    "vehicle_id"
]

df_model = df.drop(columns=drop_cols)

print("Original shape:", df.shape)
print("Modeling shape:", df_model.shape)
print("\nRemaining columns:")
print(df_model.columns.tolist())
```

```python

X = df_model.drop("breakdown_within_30days", axis=1)
y = df_model["breakdown_within_30days"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Numerical columns
numeric_cols = X_train.select_dtypes(include=np.number).columns

# Calculate IQR-based outliers
Q1 = X_train[numeric_cols].quantile(0.25)
Q3 = X_train[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_counts = (
    ((X_train[numeric_cols] < lower_bound) |
     (X_train[numeric_cols] > upper_bound))
    .sum()
    .sort_values(ascending=False)
)

print("Outlier count per feature:")
print(outlier_counts)

print("\nTotal outlier values:")
print(outlier_counts.sum())
```

```python
# Separate continuous numerical features from binary features
continuous_cols = [
    col for col in X_train.select_dtypes(include=np.number).columns
    if X_train[col].nunique() > 2
]

# Calculate IQR from training data only
Q1 = X_train[continuous_cols].quantile(0.25)
Q3 = X_train[continuous_cols].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Cap outliers in training data
X_train = X_train.copy()
X_test = X_test.copy()

for col in continuous_cols:
    X_train[col] = X_train[col].clip(
        lower=lower_bound[col],
        upper=upper_bound[col]
    )

    X_test[col] = X_test[col].clip(
        lower=lower_bound[col],
        upper=upper_bound[col]
    )

print("Outlier handling completed.")
print("Continuous features processed:", len(continuous_cols))
print("Binary features preserved:", 
      len(X_train.select_dtypes(include=np.number).columns) - len(continuous_cols))
```

```python
categorical_cols = X_train.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X_train.select_dtypes(include=[np.number]).columns.tolist()

print("Categorical features:")
print(categorical_cols)

print("\nNumerical features:")
print(numerical_cols)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=2000,
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

print("Logistic Regression trained successfully.")
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score

)

y_prob_lr = logistic_model.predict_proba(X_test_processed)[:, 1]

y_pred_lr = (y_prob_lr >= 0.5).astype(int)

print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=["No Breakdown", "Breakdown"],
        zero_division=0
    )
)

print("ROC-AUC:", round(roc_auc_score(y_test, y_prob_lr), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_prob_lr), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression

class_weights = [
    {0: 1, 1: 5},
    {0: 1, 1: 10},
    {0: 1, 1: 20},
    {0: 1, 1: 30},
    {0: 1, 1: 40},
    {0: 1, 1: 50},
    "balanced"
]

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for weight in class_weights:

    model = LogisticRegression(
        class_weight=weight,
        max_iter=2000,
        random_state=42
    )

    scores = cross_val_score(
        model,
        X_train_processed,
        y_train,
        cv=cv,
        scoring="average_precision"
    )

    results.append({
        "class_weight": str(weight),
        "mean_PR_AUC": scores.mean(),
        "std_PR_AUC": scores.std()
    })

weight_results = pd.DataFrame(results)

print(weight_results.sort_values(
    "mean_PR_AUC",
    ascending=False
))
```

```python
best_class_weight = {0: 1, 1: 5}

tuned_lr = LogisticRegression(
    class_weight=best_class_weight,
    max_iter=2000,
    random_state=42
)

tuned_lr.fit(X_train_processed, y_train)

print("Tuned Logistic Regression trained.")
print("Class weight:", best_class_weight)
```

```python

y_prob = tuned_lr.predict_proba(X_test_processed)[:, 1]

print("First 10 predicted probabilities:")
print(y_prob[:10])
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

threshold_results = []

for threshold in np.arange(0.05, 0.96, 0.05):

    y_pred = (y_prob >= threshold).astype(int)

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": round(
            precision_score(y_test, y_pred, zero_division=0), 4
        ),
        "Recall": round(
            recall_score(y_test, y_pred, zero_division=0), 4
        ),
        "F1": round(
            f1_score(y_test, y_pred, zero_division=0), 4
        )
    })

threshold_df = pd.DataFrame(threshold_results)

display(threshold_df)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_prob = cross_val_predict(
    tuned_lr,
    X_train_processed,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]

threshold_results = []

for threshold in np.arange(0.05, 0.96, 0.05):

    y_pred = (cv_prob >= threshold).astype(int)

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": round(
            precision_score(y_train, y_pred, zero_division=0), 4
        ),
        "Recall": round(
            recall_score(y_train, y_pred, zero_division=0), 4
        ),
        "F1": round(
            f1_score(y_train, y_pred, zero_division=0), 4
        )
    })

threshold_df = pd.DataFrame(threshold_results)

display(
    threshold_df.sort_values(
        "F1",
        ascending=False
    )
)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score
)

best_threshold = 0.50

y_pred_final = (y_prob >= best_threshold).astype(int)

print("Final Threshold:", best_threshold)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_final,
        target_names=["No Breakdown", "Breakdown"],
        zero_division=0
    )
)

print("ROC-AUC:", round(roc_auc_score(y_test, y_prob), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_prob), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_final))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight={0: 1, 1: 5},
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)

print("Random Forest trained successfully.")
```

```python
from sklearn.model_selection import StratifiedKFold, GridSearchCV

rf_param_grid = {
    "n_estimators": [200, 300],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "class_weight": [
        {0: 1, 1: 5},
        {0: 1, 1: 10},
        {0: 1, 1: 20}
    ]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_grid=rf_param_grid,
    scoring="average_precision",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

rf_grid.fit(X_train_processed, y_train)

print("Best Random Forest parameters:")
print(rf_grid.best_params_)

print("\nBest CV PR-AUC:")
print(round(rf_grid.best_score_, 4))
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score

best_rf = rf_grid.best_estimator_

rf_cv_prob = cross_val_predict(
    best_rf,
    X_train_processed,
    y_train,
    cv=cv,
    method="predict_proba"
)[:, 1]

rf_threshold_results = []

for threshold in np.arange(0.05, 0.96, 0.05):

    y_pred = (rf_cv_prob >= threshold).astype(int)

    rf_threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": round(
            precision_score(y_train, y_pred, zero_division=0), 4
        ),
        "Recall": round(
            recall_score(y_train, y_pred, zero_division=0), 4
        ),
        "F1": round(
            f1_score(y_train, y_pred, zero_division=0), 4
        )
    })

rf_threshold_df = pd.DataFrame(rf_threshold_results)

display(
    rf_threshold_df.sort_values(
        "F1",
        ascending=False
    )
)
```

```python
rf_final_threshold = 0.35

y_prob_rf = best_rf.predict_proba(X_test_processed)[:, 1]

y_pred_rf = (y_prob_rf >= rf_final_threshold).astype(int)

print("Random Forest Final Threshold:", rf_final_threshold)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=["No Breakdown", "Breakdown"],
        zero_division=0
    )
)

print("ROC-AUC:", round(roc_auc_score(y_test, y_prob_rf), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_prob_rf), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(),
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train_processed, y_train)

print("XGBoost trained successfully.")
```

```python
y_prob_xgb = xgb_model.predict_proba(X_test_processed)[:, 1]

y_pred_xgb = (y_prob_xgb >= 0.5).astype(int)

print("XGBoost Threshold: 0.50")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_xgb,
        target_names=["No Breakdown", "Breakdown"],
        zero_division=0
    )
)

print("ROC-AUC:", round(roc_auc_score(y_test, y_prob_xgb), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_prob_xgb), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
thresholds = np.arange(0.05, 0.96, 0.05)

threshold_results = []

for threshold in thresholds:
    y_pred = (y_prob_xgb >= threshold).astype(int)

    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": round(precision, 4),
        "Recall": round(recall, 4),
        "F1": round(f1, 4)
    })

threshold_df_xgb = pd.DataFrame(threshold_results)

display(
    threshold_df_xgb.sort_values(
        "F1",
        ascending=False
    )
)
```

```python
xgb_final_threshold = 0.70

y_pred_xgb_final = (y_prob_xgb >= xgb_final_threshold).astype(int)

print("XGBoost Final Threshold:", xgb_final_threshold)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_xgb_final,
        target_names=["No Breakdown", "Breakdown"],
        zero_division=0
    )
)

print("ROC-AUC:", round(roc_auc_score(y_test, y_prob_xgb), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_prob_xgb), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_xgb_final))
```
